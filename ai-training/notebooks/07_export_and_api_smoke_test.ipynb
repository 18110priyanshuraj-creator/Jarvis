# Adapter manifest and endpoint smoke test

Do not deploy from this notebook. First verify that adapter files, training metrics, and held-out evaluation exist. A manifest starts with deployment disabled.

```python
from pathlib import Path
import hashlib,json
adapter=Path('')
assert str(adapter)!='.' and adapter.exists(), 'Set adapter to the verified Drive adapter directory.'
manifest={'status':'not-approved','deployment_enabled':False,'files':[{'path':str(p.relative_to(adapter)),'sha256':hashlib.sha256(p.read_bytes()).hexdigest(),'bytes':p.stat().st_size} for p in sorted(adapter.rglob('*')) if p.is_file()]}
out=Path('release-manifest.json');out.write_text(json.dumps(manifest,indent=2));print(out.resolve())
```
