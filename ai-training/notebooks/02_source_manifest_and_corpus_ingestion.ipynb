# Approved source manifest and corpus review

Only ingest material you own, have documented permission to use, or whose license clearly allows the intended training use. This notebook does not scrape websites, download videos, or bypass source restrictions.

```python
from pathlib import Path
import json
schema=json.loads(Path('ai-training/schemas/source-manifest.schema.json').read_text(encoding='utf-8'))
print('Manifest schema loaded:',schema.get('$id','(no id)'))
print('Add an approved manifest entry and evidence before including any source.')
```

Review each source manually: exact permission/license, intended training purpose, privacy, acquisition method, attribution, retention, and SHA-256. Put rejected or unknown sources aside. Keep raw media separate from derived notes and examples.

```python
from jsonschema import validate
manifest_path=Path('ai-training/source_manifests/manifest.json')
if manifest_path.exists():
    manifest=json.loads(manifest_path.read_text(encoding='utf-8'))
    validate(manifest,schema)
    print('Manifest schema valid. Human approval and permission evidence still need review.')
else:
    print('No local manifest yet. Create one from the schema after reviewing sources.')
```
