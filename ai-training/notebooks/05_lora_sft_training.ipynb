Install the pinned-compatible Colab dependencies only after the runtime audit. The runner refuses CPU-only or under-sized GPUs and requires non-empty train and validation sets plus an immutable model revision.

```python
import subprocess,sys
subprocess.run([sys.executable,'-m','pip','install','-q','-r','ai-training/requirements-colab.txt'],check=True)
```

Set MODEL_ID and MODEL_REVISION to a reviewed model and immutable SHA. Use Drive train and validation files. Checkpoints are kept under the run directory and the --resume option continues the latest checkpoint.
